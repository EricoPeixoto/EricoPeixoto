Uso do spaCy no Python

```python
import spacy

nlp = spacy.load("en_core_web_sm")
texto = "Eu uso o framework spaCy."
doc = nlp(texto)
for token in doc:
  print(token.text)
```

```python
for token in doc:
  print(token.text, token.pos_)
```

```python
for ent in doc.ents:
  print(ent.text, ent.label_)
```
