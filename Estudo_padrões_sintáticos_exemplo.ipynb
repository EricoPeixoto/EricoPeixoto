Padrão sintatico exemplo usando linguagem C - base de estudo

```python
import nltk
nltk.download('punkt_tab')
```

```python
import re
from nltk.tokenize import word_tokenize

c_codigo = '''
#include <stdio.h>

int main() {
    int num = 1;
    while (num <= 10) {
        printf("%d ", num);
        num++;
    }
    return 0;
}
'''

# Tokenizar o código em palavras
tokens = word_tokenize(c_codigo)

# Buscar pelo padrão sintático do comando "while"
padrao = r"while\s*\([^)]+\)\s*\{[^}]*\}"
comando_while = re.findall(padrao, ' '.join(tokens))

for s in comando_while:
    print(s)
```
